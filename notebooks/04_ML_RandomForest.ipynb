# Bemorga dori tavsiya qilish tizimi - Random Forest Classifier

Ushbu notebookda **Random Forest** ansambl algoritm yordamida bemorlarga dori tavsiya qilish modelini quramiz va natijalarni **Decision Tree** modeli bilan solishtiramiz.

## 1. Kerakli kutubxonalarni chaqirib olamiz

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
import os
import joblib

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

## 2. Datasetni yuklash va tayyorlash

```python
sys.path.append('../src')
from preprocessing import DrugDataPreprocessor

df = pd.read_csv('../data/raw/drug200.csv')
preprocessor = DrugDataPreprocessor()
X, y = preprocessor.fit_transform(df)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

## 3. Random Forest Modelini O'rgatish

```python
rf_model = RandomForestClassifier(n_estimators=100, criterion='entropy', random_state=42)
rf_model.fit(X_train, y_train)

rf_preds = rf_model.predict(X_test)
print(f"Random Forest Test Anqiqligi: {accuracy_score(y_test, rf_preds)*100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test, rf_preds))
```

## 4. Model Visualizatsiyasi va Taqqoslash

```python
dt_model = DecisionTreeClassifier(criterion='entropy', max_depth=4, random_state=42)
dt_model.fit(X_train, y_train)

dt_scores = cross_val_score(dt_model, X, y, cv=5)
rf_scores = cross_val_score(rf_model, X, y, cv=5)

print(f"Decision Tree 5-Fold CV Anqiqligi: {dt_scores.mean()*100:.2f}%")
print(f"Random Forest 5-Fold CV Anqiqligi: {rf_scores.mean()*100:.2f}%")
```

## 5. Modelni Saqlash

```python
joblib.dump(rf_model, '../models/RandomForest.pkl')
```
